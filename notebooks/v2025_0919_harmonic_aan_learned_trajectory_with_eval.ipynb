```python
from dataclasses import dataclass
from typing import Any, Callable, Sequence, Optional, List, Tuple, Dict
@dataclass
class HarmonicParams:
    n: int
    m: np.ndarray         # (n,) masses per oscillator
    k_wall: float         # wall spring constant
    k_pair: float         # pairwise coupling spring constant
    A_modes: np.ndarray   # (n,) amplitudes per normal mode
    phi_modes: np.ndarray # (n,) phases per normal mode (radians)

def sample_harmonic_params(
    n: int,
    m_range: Tuple[float, float] = (1.0, 5.0),
    k_wall: float = 0.01,
    k_pair: float = 0.005,
    A_range: Tuple[float, float] = (1.0, 3.0),
    seed: Optional[int] = 0,
) -> HarmonicParams:
    rng = np.random.default_rng(seed)
    m = rng.uniform(*m_range, size=(n,)).astype(np.float64)
    # Normal-mode amplitudes/phases
    A_modes = rng.uniform(*A_range, size=(n,)).astype(np.float64)
    phi_modes = rng.uniform(0.0, 2*np.pi, size=(n,)).astype(np.float64)
    return HarmonicParams(
        n=n, m=m, k_wall=float(k_wall), k_pair=float(k_pair),
        A_modes=A_modes, phi_modes=phi_modes
    )

def harmonic_hamiltonian(q: np.ndarray, p: np.ndarray, params: HarmonicParams) -> float:
    """
    q, p: (B, n)
    """
    n = params.n
    V_wall = 0.5 * params.k_wall * np.sum(q**2, axis=1)  # (B,)
    V_pair = 0.5 * params.k_pair * ((np.sum(q, axis=1)**2 - np.sum(q**2, axis=1))/2.0)  # (B,)
    V = V_wall + V_pair # (B,)
    T = 0.5 * np.sum((p**2) / params.m, axis=1)  # (B,)
    H = T + V
    return float(H)

def _normal_mode_eigendecomposition(params: HarmonicParams):
    """
    Returns eigenvalues/eigenvectors for q¨ = M^{-1} L q.
    Repoに合わせて L = -(k_wall + n*k_pair) I + k_pair 1 1^T を用い、
    固有値は負になる（w = sqrt(-eigvals)）。
    """
    n = params.n
    I = np.eye(n)
    J = np.ones((n, n))
    L = -(params.k_wall + n * params.k_pair) * I + params.k_pair * J
    M_sqrt_inv = np.diag(1.0 / np.sqrt(params.m))

    # 対称化
    S = M_sqrt_inv @ L @ M_sqrt_inv
    eigvals, U = np.linalg.eigh(S)              # 決定的になりやすい
    V = M_sqrt_inv @ U                          # 元の一般化固有ベクトルへ
    return eigvals.real, V.real                 # (n,), (n,n)

    # M_inv = np.diag(1.0 / params.m)
    # M_inv_L = M_inv @ L
    # eigvals, eigvecs = np.linalg.eig(M_inv_L)
    # eigvals = eigvals.real
    # eigvecs = eigvecs.real
    # return eigvals, eigvecs  # (n,), (n,n)

def simulate_coupled_1d_harmonic(
    params: HarmonicParams, num_steps: int, dt: float
) -> Tuple[np.ndarray, np.ndarray, np.ndarray, Dict]:
    """
    Returns:
      t: (T,)
      q: (T, n) positions
      p: (T, n) momenta  (p = m * dq/dt)
      aux: dict with eigvals/eigvecs/w
    """
    n = params.n
    t = np.arange(num_steps, dtype=np.float64) * float(dt)

    # Normal modes
    eigvals, V = _normal_mode_eigendecomposition(params)
    w = np.sqrt(np.clip(-eigvals, a_min=0.0, a_max=None))  # angular freq

    # Mode-space trajectories
    Wt_phi = t[:, None] * w[None, :] + params.phi_modes[None, :]  # (T, n)
    q_mode = params.A_modes[None, :] * np.cos(Wt_phi)              # (T, n)
    qdot_mode = -params.A_modes[None, :] * (w[None, :]) * np.sin(Wt_phi)  # (T, n)

    # Map back to physical coords: q = V q_mode, qdot = V qdot_mode
    q = q_mode @ V.T
    qdot = qdot_mode @ V.T
    p = qdot * params.m[None, :]  # momentum = m * qdot

    aux = {"eigvals": eigvals, "eigvecs": V, "w": w}
    return t, q, p, aux
```

## ✅ Ground-truth Hamiltonian & Action–Angle helpers

This section provides utilities to compute:

- Ground-truth Hamiltonian  
- Action–Angle variables (I, θ) via the mass-weighted eigendecomposition

It assumes you already simulated `(t, q, p, aux)` and have `params` with fields:
`n`, `m`, `k_wall`, `k_pair`. If not, adapt the cell accordingly.

```python
import numpy as np

def make_m_k(sim_params):
    """
    Returns:
        M: (n,n) mass matrix (diag)
        K: (n,n) stiffness matrix for potential U = 1/2 q^T K q
    """
    n = int(sim_params.n) if hasattr(sim_params, "n") else int(sim_params["n"])
    m = np.asarray(getattr(sim_params, "m"), dtype=float)
    k_wall = float(getattr(sim_params, "k_wall"))
    k_pair = float(getattr(sim_params, "k_pair"))
    M = np.diag(m)
    J = np.ones((n, n))
    K = k_wall * np.eye(n) + k_pair * (n * np.eye(n) - J)
    return M, K

def H_gt(q, p, sim_params):
    """
    Ground-truth Hamiltonian evaluated pointwise.
    Args:
        q: (T,n) or (N,n)
        p: (T,n) or (N,n)
    Returns:
        H: (T,) or (N,)
    """
    M, K = make_m_k(sim_params)
    Minv = np.linalg.inv(M)
    q = np.atleast_2d(q)
    p = np.atleast_2d(p)
    T = 0.5 * np.einsum("ti,ij,tj->t", p, Minv, p)
    U = 0.5 * np.einsum("ti,ij,tj->t", q, K, q)
    return T + U

def action_angle_gt(q, p, sim_params):
    """
    Compute (I, theta) from (q, p) using mass-weighted eigendecomposition.
    Args:
        q, p: (..., n)
    Returns:
        I: (..., n)
        theta: (..., n) wrapped to (-pi, pi]
        omega: (n,)
        Umat: (n, n) eigenvectors used (orthogonal)
    """
    n = int(sim_params.n)# if hasattr(sim_params, "n") else int(sim_params["n"])
    m = np.asarray(getattr(sim_params, "m"), dtype=float)
    M = np.diag(m)
    _, K = make_m_k(sim_params)

    Minv_sqrt = np.diag(1.0 / np.sqrt(m))
    A = Minv_sqrt @ K @ Minv_sqrt
    w2, U = np.linalg.eigh(A)
    omega = np.sqrt(np.clip(w2, a_min=0.0, a_max=None))

    Msqrt = np.diag(np.sqrt(m))
    y = (Msqrt @ np.atleast_2d(q).T).T
    pi = (Minv_sqrt @ np.atleast_2d(p).T).T
    xi = (U.T @ y.T).T
    rho = (U.T @ pi.T).T

    omega_safe = np.where(omega > 0, omega, 1.0)
    I = 0.5 * ((rho**2) + (omega[None, :]**2) * (xi**2)) / omega_safe[None, :]
    theta = np.arctan2(omega[None, :] * xi, rho)
    return I, theta, omega, U
```

## 🔌 (Optional) Model hooks

- `predict_H(q, p) -> ndarray shape (T,)`
- `predict_I_theta(q, p) -> (I_pred, theta_pred)` both shape `(T, n)`

```python
# --- Predicted quantities from the AAN model ---
def predict_H(q, p):
    
    q_j = jnp.asarray(q)
    p_j = jnp.asarray(p)
    if q_j.ndim == 1:
        q_j = q_j[None, :]
    if p_j.ndim == 1:
        p_j = p_j[None, :]
    H_pred = model.apply(
        {'params': params}, 
        q_j, p_j,
        method=MyActionAngleNetwork.hamiltonian
        )

    return np.asarray(H_pred)

def predict_I_theta(q, p):
    # Compute (I, theta) implied by the encoder at current (q,p).
    # We call the model with delta_t = 0.0 to get current (I, theta).
    q_j = jnp.asarray(q)
    p_j = jnp.asarray(p)
    if q_j.ndim == 1:
        q_j = q_j[None, :]
    if p_j.ndim == 1:
        p_j = p_j[None, :]
    _, _, I, theta = model.apply({'params': params}, q_j, p_j, jnp.asarray(0.0), train=False)
    return np.asarray(I), np.asarray(theta)
```

## 🔭 Single-mode grid sweep (diagnostics)

We probe a **single normal mode** plane spanned by \((\xi_j, \rho_j)\) while keeping other modes at zero to visualize:
- Ground-truth \(H\) over a grid
- (Optionally) predicted \(H\) and their difference

**How it works**
1. Choose a mode index `j_mode`.
2. Build a grid of \((\xi_j, \rho_j)\).
3. Map to \((q, p)\) via \( q = M^{-1/2} U\,\xi,\ \ p = M^{1/2} U\,\rho \).
4. Evaluate and visualize.

```python
# Configure the grid
j_mode = 0           # which normal mode to probe
grid_pts = 61        # resolution per axis
xi_max = 2.0
rho_max = 2.0

try:
    omega = omega_true
    U = U_true
except NameError:
    _, _, omega, U = action_angle_gt(np.zeros((1, params.n)), np.zeros((1, params.n)), params)

n = int(params.n) if hasattr(params, "n") else int(params["n"])
m = np.asarray(getattr(params, "m", params["m"]), dtype=float)
Msqrt = np.diag(np.sqrt(m))

xi_lin = np.linspace(-xi_max, xi_max, grid_pts)
rho_lin = np.linspace(-rho_max, rho_max, grid_pts)

H_map = np.zeros((grid_pts, grid_pts))

for a, xi_j in enumerate(xi_lin):
    for b, rho_j in enumerate(rho_lin):
        xi = np.zeros((n,))
        rho = np.zeros((n,))
        xi[j_mode] = xi_j
        rho[j_mode] = rho_j

        y = U @ xi
        pi = U @ rho
        q_pt = np.linalg.inv(Msqrt) @ y
        p_pt = Msqrt @ pi

        H_val = H_gt(q_pt[None, :], p_pt[None, :], params)[0]
        H_map[a, b] = H_val

import matplotlib.pyplot as plt

plt.figure()
plt.imshow(H_map.T, origin="lower",
           extent=[xi_lin[0], xi_lin[-1], rho_lin[0], rho_lin[-1]],
           aspect="auto")
plt.xlabel(f"xi_{j_mode}")
plt.ylabel(f"rho_{j_mode}")
plt.title(f"Ground-truth H on mode-{j_mode} plane")
plt.colorbar()
plt.show()
```
